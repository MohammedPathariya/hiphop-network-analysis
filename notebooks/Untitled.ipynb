```python
from PIL import Image
import os

# Define your paths (Adjust these to match your folder structure)
# Assuming your images are in '../output/figures' based on previous notebooks
input_dir = '../output/figures' 
output_dir = '../output/figures_pdf'

# Create output folder if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

print(f"🚀 Converting images from {input_dir} to PDF...")

for filename in os.listdir(input_dir):
    if filename.lower().endswith((".png", ".jpg", ".jpeg")):
        try:
            # Construct paths
            img_path = os.path.join(input_dir, filename)
            pdf_filename = os.path.splitext(filename)[0] + ".pdf"
            output_path = os.path.join(output_dir, pdf_filename)

            with Image.open(img_path) as image:
                # convert('RGB') is needed because some PNGs have transparency (RGBA)
                # which isn't compatible with the basic PDF standard
                rgb_im = image.convert('RGB')
                
                # Save as PDF with high resolution
                rgb_im.save(output_path, "PDF", resolution=300.0)
                print(f"✅ Converted: {filename} -> {pdf_filename}")
                
        except Exception as e:
            print(f"❌ Error converting {filename}: {e}")

print("\n🎉 All done! Upload the files in 'figures_pdf' to Overleaf.")
```
